# Gayanara E-commerce: Insights dan Rekomendasi
## Ringkasan
Periode Januari 2022 sampai Februari 2025: revenue Rp.1,27 miliar dari 2.521 order valid, rata-rata Rp.501.920 per order. 16% order dibatalkan atau dikembalikan.

```python
import pandas as pd
orders = pd.read_csv("data_clean/orders_clean.csv", parse_dates=["order_date"])
valid = orders[orders["is_valid_revenue"]]
print("Revenue:", valid["total_amount_idr"].sum())
print("Order  Valid:", len(valid))
print("AOV:", round(valid["total_amount_idr"].mean()))
print("Rate gagal (%):", round((~orders["is_valid_revenue"]).mean() * 100, 1))
```

## Temuan Utama
1. **Revenue tumbuh.** Naik tajam di awal 2023 dan trennya naik bertahap sampai puncak Februari 2025 (Rp.66,3 juta).
2. **Kategori merata.** Lima kategori berselisih tipis (Rp.216 sampai 224 juta). T-Shirt paling rendah (Rp.195,6 juta).
3. **Kota merata.** 20 kota, tidak ada yang dominan. Semarang tertinggi (Rp.76,6 juta).
4. **Promo tidak menaikkan nilai order.** 41% order pakai promo, tapi rata-rata nilainya sekotar 6% lebih kecil.
5. **Order gagal 16%.** J&T (18,4%) dan OVO (18,7%) tertinggi, selisihnya kecil.

## Rekomendasi
1. Evaluasi program promo, karena diskonnya kecil (2,4% revenue) tapi tidak terlihat menaikkan nilai order.
2. Cari tau penyebab rendahnya T-Shirt (harga, stok, atau tampilan produk).
3. Pantau order gagal per kurir dan metode bayar sebelum mengambil keputusan.
